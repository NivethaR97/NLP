```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import f1_score
from sklearn.model_selection import GridSearchCV
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("whatsapp_chat_labeled.csv")
dataset
```

```python
x=dataset["Chat"]
y=dataset["Label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=53)
```

```python
hash_vectorizer=HashingVectorizer(n_features=10, norm=None, alternate_sign=False,stop_words='english')
hash_train = hash_vectorizer.fit_transform(X_train)
hash_test = hash_vectorizer.transform(X_test)
print(hash_train.toarray())
```

```python
clf = DecisionTreeClassifier(class_weight= 'balanced',max_features= 'sqrt',min_samples_split=3,splitter= 'random',random_state=42)
clf.fit(hash_train, y_train)
pred = clf.predict(hash_test)
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)
cm = metrics.confusion_matrix(y_test, pred)
print("The confusion Matrix:\n",cm)
clf_report = classification_report(y_test, pred)
print("The report:\n",clf_report)
```
