```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier,LogisticRegression
from sklearn.naive_bayes import MultinomialNB,GaussianNB,ComplementNB,BernoulliNB,CategoricalNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn import metrics
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("whatsapp_chat_labeled.csv")
dataset
```

```python
x=dataset["Chat"]
y=dataset["Label"]
```

```python
def  count_split(x,y):
    X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=53)
    count_vectorizer = CountVectorizer(stop_words='english')
    count_train = count_vectorizer.fit_transform(X_train)
    count_test = count_vectorizer.transform(X_test)
    len(count_vectorizer.get_feature_names_out())#unique words
    print(count_train.toarray())
    return count_train,count_test,y_train,y_test
```

```python
def cm_prediction(classifier,X_test,y_test):
    y_pred=classifier.predict(X_test)
    from sklearn.metrics import confusion_matrix
    cm=confusion_matrix(y_test,y_pred)
    from sklearn.metrics import accuracy_score
    from sklearn.metrics import classification_report
    Accuracy=accuracy_score(y_test,y_pred)
    report=classification_report(y_test,y_pred)
    from sklearn.metrics import f1_score
    f1=f1_score(y_test,y_pred,pos_label='Flirty')
    return Accuracy,report,cm,f1
```

```python
def Multinomial(count_train,y_train,count_test,y_test):
    classifier = MultinomialNB()
    classifier.fit(count_train, y_train)
    pred = classifier.predict(count_test)
    score = metrics.accuracy_score(y_test, pred)
    Accuracy,report,cm,f1=cm_prediction(classifier,count_test,y_test)
    return  classifier,Accuracy,report,cm,f1
```

```python
def logistic(count_train,y_train,count_test,y_test):
    from sklearn.linear_model import LogisticRegression
    classifier=LogisticRegression(random_state=0,max_iter=1000)
    classifier.fit(count_train,y_train)
    Accuracy,report,cm,f1=cm_prediction(classifier,count_test,y_test)
    return classifier,Accuracy,report,cm,f1  
```

```python
def svm_linear(count_train,y_train,count_test,y_test):
    from sklearn.svm import SVC
    classifier = SVC(kernel = 'linear', random_state = 0)
    classifier.fit(count_train, y_train)
    Accuracy,report,cm,f1=cm_prediction(classifier,count_test,y_test)
    return  classifier,Accuracy,report,cm,f1
```

```python
def svm_NL(count_train,y_train,count_test,y_test):
    from sklearn.svm import SVC
    classifier = SVC(kernel = 'rbf', random_state = 0)
    classifier.fit(count_train, y_train)
    Accuracy,report,cm,f1=cm_prediction(classifier,count_test,y_test)
    return  classifier,Accuracy,report,cm,f1
```

```python
def knn(count_train,y_train,count_test,y_test):
    from sklearn.neighbors import KNeighborsClassifier
    classifier = KNeighborsClassifier(n_neighbors = 5, metric = 'minkowski', p = 2)
    classifier.fit(count_train, y_train)
    Accuracy,report,cm,f1=cm_prediction(classifier,count_test,y_test)
    return classifier,Accuracy,report,cm,f1
```

```python
def Decision(count_train,y_train,count_test,y_test):
    from sklearn.tree import DecisionTreeClassifier
    classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
    classifier.fit(count_train, y_train)
    Accuracy,report,cm,f1=cm_prediction(classifier,count_test,y_test)
    return classifier,Accuracy,report,cm,f1     
```

```python
def random(count_train,y_train,count_test,y_test):
    from sklearn.ensemble import RandomForestClassifier
    classifier = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 0)
    classifier.fit(count_train, y_train)
    Accuracy,report,cm,f1=cm_prediction(classifier,count_test,y_test)
    return classifier,Accuracy,report,cm,f1
```

```python
accnav=[]
acclog=[]
accsvml=[]
accsvmnl=[]
accknn=[]
accdes=[]
accrf=[]

count_train, count_test, y_train, y_test = count_split(x,y)
classifier,Accuracy,report,cm,f1=Multinomial(count_train,y_train,count_test,y_test)  
accnav.append(Accuracy)

classifier,Accuracy,report,cm,f1=logistic(count_train,y_train,count_test,y_test)
acclog.append(Accuracy)

classifier,Accuracy,report,cm,f1=svm_linear(count_train,y_train,count_test,y_test)  
accsvml.append(Accuracy)
    
classifier,Accuracy,report,cm,f1=svm_NL(count_train,y_train,count_test,y_test)  
accsvmnl.append(Accuracy)
    
classifier,Accuracy,report,cm,f1=knn(count_train,y_train,count_test,y_test)  
accknn.append(Accuracy)
    
classifier,Accuracy,report,cm,f1=Decision(count_train,y_train,count_test,y_test)  
accdes.append(Accuracy)
    
classifier,Accuracy,report,cm,f1=random(count_train,y_train,count_test,y_test)  
accrf.append(Accuracy)


```

```python
def Flirt_Classification(accnav,acclog,accsvml,accsvmnl,accknn,accdes,accrf):
    dataframe=pd.DataFrame(index=['Accuracy'],columns=['Mutinomial','Logistic','SVMl','SVMnl','KNN','Decision','Random'])
    dataframe.loc['Accuracy','Multinomial']=accnav[0]
    dataframe.loc['Accuracy','Logistic']=acclog[0]
    dataframe.loc['Accuracy','SVMl']=accsvml[0]
    dataframe.loc['Accuracy','SVMl']=accsvml[0]
    dataframe.loc['Accuracy','SVMnl']=accsvmnl[0]
    dataframe.loc['Accuracy','KNN']=accknn[0]
    dataframe.loc['Accuracy','Decision']=accdes[0]
    dataframe.loc['Accuracy','Random']=accrf[0]
    return dataframe      
```

```python
result=Flirt_Classification(accnav,acclog,accsvml,accsvmnl,accknn,accdes,accrf)
result
```
