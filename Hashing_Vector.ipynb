```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier,LogisticRegression
from sklearn.naive_bayes import MultinomialNB,GaussianNB,ComplementNB,BernoulliNB,CategoricalNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn import metrics
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("whatsapp_chat_labeled.csv")
dataset
```

```python
x=dataset["Chat"]
y=dataset["Label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=53)
```

```python
hash_vectorizer=HashingVectorizer(n_features=10, norm=None, alternate_sign=False,stop_words='english')
hash_train = hash_vectorizer.fit_transform(X_train)
hash_test = hash_vectorizer.transform(X_test)
print(hash_train.toarray())
```

```python
classifier = MultinomialNB()
classifier.fit(hash_train, y_train)
pred = classifier.predict(hash_test) 
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)
cm = metrics.confusion_matrix(y_test, pred)
print("confusion_matrix:\n",cm)
report=classification_report(y_test, pred)
print(report)
```

```python
classifier=LogisticRegression(random_state=0,max_iter=1000)
classifier.fit(hash_train,y_train)
pred = classifier.predict(hash_test) 
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)
cm = metrics.confusion_matrix(y_test, pred)
print("confusion_matrix:\n",cm)
report=classification_report(y_test, pred)
print(report) 
```

```python
classifier = SVC(kernel = 'linear', random_state = 0)
classifier.fit(hash_train, y_train)
pred = classifier.predict(hash_test) 
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)
cm = metrics.confusion_matrix(y_test, pred)
print("confusion_matrix:\n",cm)
report=classification_report(y_test, pred)
print(report) 
```

```python
classifier = KNeighborsClassifier(n_neighbors = 5)
classifier.fit(hash_train, y_train)
pred = classifier.predict(hash_test) 
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)
cm = metrics.confusion_matrix(y_test, pred)
print("confusion_matrix:\n",cm)
report=classification_report(y_test, pred)
print(report) 
```

```python
classifier = DecisionTreeClassifier(min_samples_split=3, random_state = 0,class_weight='balanced')
classifier.fit(hash_train, y_train)
pred = classifier.predict(hash_test) 
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)
cm = metrics.confusion_matrix(y_test, pred)
print("confusion_matrix:\n",cm)
report=classification_report(y_test, pred)
print(report)  
```

```python
classifier = RandomForestClassifier(n_estimators =50,min_samples_split=3,class_weight="balanced")
classifier.fit(hash_train, y_train)
pred = classifier.predict(hash_test) 
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)
cm = metrics.confusion_matrix(y_test, pred)
print("confusion_matrix:\n",cm)
report=classification_report(y_test, pred)
print(report)  
```

```python
classifier=PassiveAggressiveClassifier(C=0.1)
classifier.fit(hash_train, y_train)
pred = classifier.predict(hash_test) 
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)
cm = metrics.confusion_matrix(y_test, pred)
print("confusion_matrix:\n",cm)
report=classification_report(y_test, pred)
print(report)  
```
